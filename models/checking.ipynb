```python
import pickle
from rdkit import Chem
from rdkit.Chem import AllChem
import joblib


# Loading the fingerprint configuration and the trained model
with open("fingerprint_config.pkl", "rb") as f:
    fp_config = pickle.load(f)

# Load the model with joblib instead of pickle


rf_model = joblib.load("ames_random_forest.pkl")

# Featurizing a SMILES string using the loaded config
#Trying out for a non toxic molecule: Aspirin
smiles = "CC(=O)OC1=CC=CC=C1C(=O)O"  

mol = Chem.MolFromSmiles(smiles)

# Generating Morgan Fingerprint as defined by config
fp = AllChem.GetMorganFingerprintAsBitVect(
    mol, radius=fp_config["radius"], nBits=fp_config["n_bits"]
)

#Predict Ames Toxicity Label (0 or 1)
toxicity_label = rf_model.predict([list(fp)])[0]
print("Toxicity Label:", toxicity_label)

```
