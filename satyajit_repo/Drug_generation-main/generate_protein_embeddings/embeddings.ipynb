### **Embeddings function from bio_embeddings**

```python
import bio_embeddings
from bio_embeddings.embed import ProtTransBertBFDEmbedder

def give_bioembeddings(protein_list):
    """
    List(str): Takes list of fasta sequences of protein 

    returns:
    List of protein embeddings. 
    """
    embedder = ProtTransBertBFDEmbedder()
    embedding_list = []
    for protein in protein_list:
        embedding_protein = embedder.embed(protein)
        embedding_per_protein = embedder.reduce_per_protein(embedding_protein)
        embedding_list.append(embedding_per_protein)
    print("\nEmbedding for this batch is done")    
    return embedding_list
```

#### Load list of fasta sequences (we are currently doing 10)

```python
import os

fasta_dir = "/home/muhamed/repo/CTDDG/datasets/data/bindingdb/test_dataset/test_4_org_1042_104"

fasta_ls = []

for i in range(1, 11):
    fasta_path = os.path.join(fasta_dir, f"protein{i}.fasta")

    with open(fasta_path, "r") as f:
        lines = f.readlines()

    # Remove FASTA header and join sequence lines
    sequence = "".join(
        line.strip()
        for line in lines
        if not line.startswith(">")
    )

    fasta_ls.append(sequence)

print("Number of proteins loaded:", len(fasta_ls))

embeds = give_bioembeddings(fasta_ls)
```

#### Save embeddings in a txt file

```python
import os

output_dir = "/home/muhamed/repo/CTDDG/ctddg_results/embeddings"
os.makedirs(output_dir, exist_ok=True)

for i, embed in enumerate(embeds, 1):
    output_path = os.path.join(output_dir, f"protein{i}.txt")

    with open(output_path, "w+") as f:
        for e in embed:
            f.write(str(e) + "\t")

    print(f"Saved protein {i}")
```
